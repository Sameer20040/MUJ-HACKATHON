# 02. Lake Segmentation Model Training
Training Multi-band U-Net and RGB Fallback Models with Combined Dice + BCE Loss.

```
import torch
from src.segmentation.unet_model import LakeUNet
from src.segmentation.losses import BCEDiceLoss

model = LakeUNet.create_multiband_model(n_channels=7)
print('Model Total Parameters:', sum(p.numel() for p in model.parameters()))
x = torch.randn(2, 7, 256, 256)
logits = model(x)
print('Output Logits Shape:', logits.shape)
```

```
loss_fn = BCEDiceLoss(dice_weight=0.5, bce_weight=0.5)
targets = torch.zeros(2, 1, 256, 256)
loss = loss_fn(logits, targets)
print('Initial Loss:', loss.item())
```
