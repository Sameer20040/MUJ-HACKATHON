# 04. Cartographic Map Generation & 3-Panel Scientific Visuals
Replicating scientific literature figures (Panels a, b, c).

```
import numpy as np
import matplotlib.pyplot as plt
from src.dataset.build_input_stack import create_synthetic_glacial_scene
from src.inference.overlay import create_dual_epoch_overlay, draw_classified_markers, create_publication_three_panel_figure
from src.postprocessing.vectorize import mask_to_geodataframe
from src.change_detection.classify_change import classify_temporal_change

stack_16, mask_16, prof = create_synthetic_glacial_scene(600, 600, seed=42, epoch=2016)
stack_22, mask_22, _ = create_synthetic_glacial_scene(600, 600, seed=100, epoch=2022)

rgb_22 = np.stack([stack_22[2], stack_22[1], stack_22[0]], axis=-1)
rgb_22 = np.clip(rgb_22 * 255.0, 0, 255).astype(np.uint8)

gdf_16 = mask_to_geodataframe(mask_16, prof['transform'])
gdf_22 = mask_to_geodataframe(mask_22, prof['transform'])
_, summary_df = classify_temporal_change(gdf_16, gdf_22)

panel_b = create_dual_epoch_overlay(rgb_22, mask_16, mask_22)
panel_c = draw_classified_markers(rgb_22, summary_df.to_dict(orient='records'))
fig = create_publication_three_panel_figure(rgb_22, panel_b, panel_c, output_path='outputs/maps/three_panel_visual.png')
plt.show()
```
